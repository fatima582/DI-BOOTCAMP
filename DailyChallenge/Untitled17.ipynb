## 1. Préparation des données

```python
import numpy as np
import pandas as pd

# Générer des températures moyennes mensuelles pour 10 villes (de -5 à 35 degrés)
np.random.seed(42) # pour la reproductibilité
temperatures = np.random.uniform(low=-5, high=35, size=(10, 12))

# Noms des villes et des mois
villes = [f'Ville_{i+1}' for i in range(10)]
mois = ['Jan', 'Fév', 'Mar', 'Avr', 'Mai', 'Juin', 'Juil', 'Août', 'Sep', 'Oct', 'Nov', 'Déc']

# Créer le DataFrame Pandas
df_temperatures = pd.DataFrame(temperatures, index=villes, columns=mois)

print("DataFrame des températures généré :")
display(df_temperatures.head())
```

## 2. Analyse des données

```python
# Calculer la température moyenne annuelle pour chaque ville
df_temperatures['Moyenne Annuelle'] = df_temperatures.mean(axis=1)

# Identifier la ville avec la température moyenne la plus élevée
ville_plus_chaude = df_temperatures['Moyenne Annuelle'].idxmax()
temp_plus_chaude = df_temperatures['Moyenne Annuelle'].max()

# Identifier la ville avec la température moyenne la plus basse
ville_plus_froide = df_temperatures['Moyenne Annuelle'].idxmin()
temp_plus_froide = df_temperatures['Moyenne Annuelle'].min()

print("DataFrame avec les températures moyennes annuelles :")
display(df_temperatures[['Moyenne Annuelle']].head())

print(f"\nVille avec la température moyenne annuelle la plus élevée : {ville_plus_chaude} ({temp_plus_chaude:.2f} °C)")
print(f"Ville avec la température moyenne annuelle la plus basse : {ville_plus_froide} ({temp_plus_froide:.2f} °C)")
```

## 3. Visualisation des données

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Préparer les données pour la visualisation (enlever la colonne 'Moyenne Annuelle')
df_plot = df_temperatures.drop(columns=['Moyenne Annuelle'])

# Transposer le DataFrame pour que les mois soient sur l'axe des x
df_plot_transposed = df_plot.T

plt.figure(figsize=(14, 7))
sns.lineplot(data=df_plot_transposed, dashes=False)
plt.title('Tendances des températures moyennes mensuelles par ville')
plt.xlabel('Mois')
plt.ylabel('Température moyenne (°C)')
plt.xticks(rotation=45)
plt.legend(title='Ville', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True)
plt.tight_layout()
plt.show()
```
