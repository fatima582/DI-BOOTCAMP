```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.decomposition import PCA

# 1. Chargement des données
# Remplace 'ton_fichier.csv' par le nom exact de ton fichier téléchargé
df = pd.read_csv('datascience_salaries.csv')

print("--- Aperçu des données initiales ---")
print(df.head(), "\n")

# 2. Normalisation Min-Max de la colonne 'salary_in_usd' (ou 'salary')
# Note : Dans ces datasets, on utilise généralement 'salary_in_usd' pour uniformiser les devises.
scaler = MinMaxScaler()
# On ajuste et transforme les données en les remodelant à 2D
df['salary_normalized'] = scaler.fit_transform(df[['salary_in_usd']])

print("--- Aperçu après Normalisation ---")
print(df[['salary_in_usd', 'salary_normalized']].head(), "\n")

# 3. Réduction de dimensionnalité (ACP)
# L'ACP ne fonctionne que sur des variables numériques ou encodées.
# Sélectionnons les colonnes numériques disponibles (ex: work_year, remote_ratio, salary_normalized)
features_numeriques = df.select_dtypes(include=['int64', 'float64']).columns.tolist()

# Si des colonnes comme 'salary' ou 'salary_in_usd' faussent l'ACP car déjà normalisées/liées, on peut filtrer :
features_pour_acp = [col for col in features_numeriques if col not in ['salary', 'salary_in_usd']]

print(f"Caractéristiques utilisées pour l'ACP : {features_pour_acp}")

# Initialisation de l'ACP pour réduire à 2 composantes principales
pca = PCA(n_components=2)
composantes_principales = pca.fit_transform(df[features_pour_acp])

# Ajout des composantes au DataFrame pour une future visualisation
df['ACP_1'] = composantes_principales[:, 0]
df['ACP_2'] = composantes_principales[:, 1]

print("\n--- Aperçu après ACP ---")
print(df[['ACP_1', 'ACP_2']].head(), "\n")

# 4. Agrégation : Salaire moyen et médian par niveau d'expérience
# Le groupe se fait généralement sur la colonne 'experience_level'
agregation_salaire = df.groupby('experience_level')['salary_in_usd'].agg(['mean', 'median']).reset_index()

# Renommer les colonnes pour plus de clarté
agregation_salaire.columns = ['Niveau d_experience', 'Salaire Moyen ($)', 'Salaire Median ($)']

print("--- Analyse et Agrégation des Salaires ---")
print(agregation_salaire)
```

Ce projet présente la résolution d'un défi d'analyse de données axé sur le traitement des salaires dans les métiers de la data science, la normalisation des caractéristiques et la réduction de dimensionnalité.

##  Fonctionnalités implémentées

### 1. Collecte et Importation des données
* Chargement du jeu de données des salaires avec **Pandas**.
* Exploration initiale de la structure du dataset (colonnes, types de données, valeurs manquantes).

### 2. Normalisation Min-Max
Pour éviter que la variable `salary_in_usd` ne domine disproportionnellement les autres caractéristiques lors des futures phases d'apprentissage automatique, nous avons appliqué une normalisation Min-Max.
* **Résultat :** Toutes les valeurs de salaires ont été ramenées à une échelle stricte et harmonisée entre **0 et 1**.

### 3. Réduction de Dimensionnalité (ACP)
Afin de simplifier le jeu de données et de lutter contre le *fléau de la dimensionnalité*, une **Analyse en Composantes Principales (ACP)** a été mise en œuvre.
* Les variables numériques (comme le taux de télétravail, l'année et le salaire normalisé) ont été projetées sur **2 composantes principales** principales (`ACP_1` et `ACP_2`).
* Cela permet de condenser un maximum d'information (variance) tout en réduisant le bruit et le nombre de colonnes.
