### Tests d'hypothèse et corrélation

#### Tests d'hypothèse (exemple : ANOVA entre groupes de prix)

```python
from scipy.stats import f_oneway

# Group data by price_range for a feature, e.g., 'ram'
# We will perform an ANOVA test for 'ram' across different price ranges.

# Get unique price ranges
price_ranges = train_df['price_range'].unique()

print("Performing ANOVA tests for key features across different price ranges:")

# Example for 'ram'
ram_groups = [train_df['ram'][train_df['price_range'] == r] for r in price_ranges]
f_stat_ram, p_value_ram = f_oneway(*ram_groups)
print(f"ANOVA for 'ram' across price ranges: F-statistic = {f_stat_ram:.2f}, p-value = {p_value_ram:.3e}")

# Example for 'battery_power'
battery_power_groups = [train_df['battery_power'][train_df['price_range'] == r] for r in price_ranges]
f_stat_battery, p_value_battery = f_oneway(*battery_power_groups)
print(f"ANOVA for 'battery_power' across price ranges: F-statistic = {f_stat_battery:.2f}, p-value = {p_value_battery:.3e}")

# Example for 'px_height' (pixel height)
px_height_groups = [train_df['px_height'][train_df['price_range'] == r] for r in price_ranges]
f_stat_px_height, p_value_px_height = f_oneway(*px_height_groups)
print(f"ANOVA for 'px_height' across price ranges: F-statistic = {f_stat_px_height:.2f}, p-value = {p_value_px_height:.3e}")

# Example for 'int_memory' (internal memory)
int_memory_groups = [train_df['int_memory'][train_df['price_range'] == r] for r in price_ranges]
f_stat_int_memory, p_value_int_memory = f_oneway(*int_memory_groups)
print(f"ANOVA for 'int_memory' across price ranges: F-statistic = {f_stat_int_memory:.2f}, p-value = {p_value_int_memory:.3e}")
```

## 5. Synthèse des idées et conclusion

### Conclusions de l'analyse statistique et des visualisations

L'analyse statistique et les visualisations ont révélé plusieurs informations clés concernant les facteurs déterminant la gamme de prix des téléphones mobiles :

1.  **Mémoire vive (RAM) :** La RAM est de loin le facteur le plus influent sur la gamme de prix, comme le montrent un coefficient de corrélation de 0,92 avec `price_range` et un F-statistique ANOVA très élevé (3520.11) avec un p-value de 0.000e+00. Les box plots et les nuages de points confirment une forte augmentation de la RAM à mesure que la gamme de prix augmente.
2.  **Puissance de la batterie (`battery_power`) :** La puissance de la batterie montre une corrélation positive modérée (0,20) avec la gamme de prix. L'ANOVA a également révélé une signification statistique (F-statistique = 31.60, p-value = 5.949e-20), indiquant des différences significatives de puissance de batterie entre les gammes de prix. Les box plots montrent une tendance à l'augmentation de la puissance de la batterie avec des gammes de prix plus élevées, bien qu'il y ait un certain chevauchement.
3.  **Résolution d'affichage (`px_width`, `px_height`) :** La largeur et la hauteur des pixels (`px_width` et `px_height`) présentent des corrélations positives avec `price_range` (0,17 et 0,15 respectivement). L'ANOVA pour `px_height` (F-statistique = 19.48, p-value = 1.886e-12) confirme que la hauteur des pixels varie significativement entre les gammes de prix. Les nuages de points et les box plots illustrent une légère tendance à des résolutions d'écran plus élevées dans les téléphones plus chers.
4.  **Mémoire interne (`int_memory`) :** La mémoire interne a une corrélation positive faible (0,04) et un F-statistique ANOVA de 2.92 avec un p-value de 0.033, ce qui est significatif au seuil de 5%. Cela suggère que même si son influence est moindre, une plus grande mémoire interne tend à être associée à des gammes de prix plus élevées.
5.  **Autres caractéristiques :** La plupart des autres caractéristiques, telles que `blue` (Bluetooth), `dual_sim`, `four_g`, `three_g`, `wifi`, `fc` (caméra frontale), `sc_h` (hauteur de l'écran), `sc_w` (largeur de l'écran), `talk_time` et `pc` (caméra principale), `n_cores`, et `m_dep` (profondeur de la batterie), `clock_speed` et `mobile_wt` (poids du mobile), `touch_screen`, montrent des corrélations très faibles ou négligeables avec la gamme de prix. Leur signification statistique, bien que parfois présente à des p-values faibles en raison de la taille de l'échantillon, ne les désigne pas comme des déterminants majeurs de la classification des prix par rapport à la RAM, à la puissance de la batterie et à la résolution d'affichage.

### Principaux facteurs déterminants de la classification des prix des téléphones mobiles

Les principaux facteurs déterminants de la classification des prix des téléphones mobiles sont :

*   **RAM :** C'est le facteur le plus prédictif. Les téléphones dotés de plus de RAM sont presque systématiquement classés dans des gammes de prix plus élevées.
*   **Puissance de la batterie :** Bien que dans une moindre mesure que la RAM, une batterie plus puissante contribue également à une gamme de prix plus élevée.
*   **Résolution d'affichage (px_width, px_height) :** Une résolution d'écran plus élevée (pixels plus nombreux) tend à faire monter le prix du téléphone.

### Découvertes inattendues ou significatives

*   **Faible impact des cœurs de processeur (`n_cores`) et de la vitesse d'horloge (`clock_speed`) :** Contrairement à l'intuition courante, le nombre de cœurs de processeur et la vitesse d'horloge n'ont montré aucune corrélation significative avec la gamme de prix. Cela pourrait suggérer que la puissance de traitement seule n'est pas un facteur distinctif suffisant pour justifier une différence de prix élevée, ou que les performances sont plus complexes que la simple vitesse d'horloge et le nombre de cœurs.
*   **Poids du mobile (`mobile_wt`) et écran tactile (`touch_screen`) :** Le poids du mobile et la présence d'un écran tactile ont des corrélations légèrement négatives, indiquant qu'ils ne sont pas des moteurs de prix positifs et peuvent même avoir une association inverse (par exemple, les téléphones plus légers ou avec des technologies d'écran plus avancées peuvent être plus chers, mais la corrélation est trop faible pour être concluante sur ce dernier point). Les box plots confirment la superposition des distributions de ces caractéristiques sur les différentes gammes de prix.
*   **Importance des technologies de connectivité :** Les caractéristiques comme le `four_g`, le `three_g`, le `wifi`, et le `blue` (Bluetooth) n'ont pas une grande influence individuelle sur la gamme de prix, ce qui pourrait indiquer qu'elles sont devenues des caractéristiques standard dans toutes les gammes de prix ou que leurs variations n'ont pas un impact suffisant pour différencier les prix de manière significative.

En résumé, pour prédire la gamme de prix d'un téléphone mobile, la RAM est le facteur le plus critique, suivie par la puissance de la batterie et la résolution d'affichage. Les autres spécifications techniques, telles que le nombre de cœurs du processeur ou la présence de certaines connectivités, semblent moins déterminantes dans cet ensemble de données.
